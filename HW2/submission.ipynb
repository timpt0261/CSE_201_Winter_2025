# Questions

### Question 1
Provide a code block for a user to generate a single shortest length decomposition for \(n\).



### Question 2
Provide a code block to generate the list of all shortest length decompositions for \(n\).



### Question 3
For all \(n <= 10,000\), find the length of the shortest decomposition. Output a histogram of the numbers (just a simple list). So state how many numbers have decomposition length 1 (obviously, these are square numbers), how many have decomposition length 2, then length 3, etc. Just print out this list.

### Question 4
What did you observe? (If you did everything correctly, you would have empirically discovered a deep fact of number theory.)

### Question 5
For all \(n \leq 500\), find the number of shortest length decompositions. Plot that as a function of \(n\).

### Question 6
For each \(n \leq 1000\), get the "wall-clock" running time with the `timelt` module. Plot that as a function of \(n\). Any guesses for what the running time function is, in big-Oh notation? As you might see, it might be difficult to handle \(n = 10^6\) on a laptop.
